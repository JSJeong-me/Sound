```python
# 1. Colab에 Google Drive 마운트
from google.colab import drive
drive.mount('/content/drive')

# 2. gdown 설치 (Google Drive 파일 직접 다운로드용)
!pip install -U gdown

# 3. 파일 ID 추출 및 다운로드
import gdown

# 링크: https://drive.google.com/file/d/w7/view?usp=sharing
# file_id = '1CL'
# https://drive.google.com/file/d/1gKq/view?usp=sharing
file_id = '1CE'
gdown.download(f'https://drive.google.com/uc?id={file_id}', 'audio_file.wav', quiet=False)

# 4. 파일 경로 지정
audio_path = 'audio_file.wav'  # 다운받은 파일명

# 이제 아래와 같이 사용하면 됩니다.
print(f"Audio file path: {audio_path}")
```

```python
import whisper

# Load the Whisper model (base is fast and decent, 'large' is most accurate but slow)
model = whisper.load_model("base")

# Path to your local audio file
# audio_path = "your_file_name_here.wav"  # Update this

# Transcribe
# result = model.transcribe(audio_path)
result = model.transcribe(audio_path, language='ko')  # for Korean

# Print the text result
print(result['text'])

# (Optional) Save to a file
with open("transcription.txt", "w", encoding="utf-8") as f:
    f.write(result['text'])
```
