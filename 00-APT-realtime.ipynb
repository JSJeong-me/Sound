```python
from openai import OpenAI
from IPython.display import Audio, display
import numpy as np
import soundfile as sf
import io
import os

client = OpenAI(api_key=os.environ["OPENAI_API_KEY"])

def synthesize_tts_to_wav(text, voice="verse", sample_rate=24000, out_path="output.wav"):
    """
    OpenAI TTS (gpt-4o-mini-tts)로 텍스트를 WAV로 합성.
    voice: 'verse', 'alloy', 'shimmer' 등
    """
    # gpt-4o-mini-tts는 텍스트 → 음성 합성용 소형 모델 (저지연/저비용)
    # Colab에서는 실시간(WebRTC) 대신 파일로 받아 재생하는 방식이 안정적입니다.
    resp = client.audio.speech.create(
        model="gpt-4o-mini-tts",
        voice=voice,
        input=text,
        # Removed the 'format' parameter as it caused a TypeError
        # format="wav"       # wav/mp3 가능
    )

    # SDK가 반환하는 바이트를 파일로 저장
    if hasattr(resp, "audio"):
        audio_bytes = resp.audio
    elif hasattr(resp, "content"):
        audio_bytes = resp.content
    else:
        # 드물게 base64로 오기도 함
        import base64
        audio_bytes = base64.b64decode(resp["data"][0]["b64_audio"])

    with open(out_path, "wb") as f:
        f.write(audio_bytes)

    return out_path

wav_path = synthesize_tts_to_wav(lyrics, voice="coral", sample_rate=24000, out_path="song.wav")
# alloy, ash, ballad, coral, echo, fable, onyx, nova, sage, shimmer, verse
display(Audio(wav_path))
print(f"Saved to: {os.path.abspath(wav_path)}")
```
